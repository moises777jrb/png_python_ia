## funçoes lambda e funçoes anonimas


```python
import math

raiz = lambda x: math.sqrt(x)
raiz(10)
```

#list comprehesion

```python
squares = [x**2 for x in range(10)]
print(squares)
```

#funcoes map

```python
import math
raiz = list(map(lambda x: math.sqrt(x),squares))
print(raiz)
```
